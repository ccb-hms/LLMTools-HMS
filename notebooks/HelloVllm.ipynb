### Getting Started with VLLM ###

```python
import sys
from openai import OpenAI

%load_ext autoreload
%autoreload 2
import llmtools
print(f'Package version: {llmtools.__version__}')
print(f'Authors: {llmtools.__authors__}')
```

```python
client = OpenAI(
    base_url="http://vllm:8000/v1",  # <-- service name on the docker network
    api_key="vllmkey"             # vLLM doesn't require a real key by default
)
```

```python
resp = client.completions.create(
    model="opt",
    prompt="User: Write a short poem.\nAssistant:",
    max_tokens=800,
)
print(resp.choices[0].text)
```

```python
response = client.responses.create(
    model="opt",
    input="Write a one-sentence bedtime story about a unicorn."
)
print(response.output_text)
```
